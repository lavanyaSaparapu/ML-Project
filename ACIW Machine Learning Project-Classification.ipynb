```python
def log_transform(col):
    return np.log(col[0])

framing_df['education'] = framing_df[['education']].apply(log_transform, axis=1)
framing_df['currentSmoker'] = framing_df[['currentSmoker']].apply(log_transform, axis=1)
framing_df['prevalentStroke'] = framing_df[['prevalentStroke']].apply(log_transform, axis=1)
framing_df['heartRate'] = framing_df[['heartRate']].apply(log_transform, axis=1)
framing_df['glucose'] = framing_df[['glucose']].apply(log_transform, axis=1)
```

```python
X=framing_df.drop(columns='TenYearCHD')

y=framing_df['TenYearCHD']
```

```python
y.value_counts()
```

```python
!pip  install imblearn
```

```python
from imblearn.over_sampling import SMOTE
oversample=SMOTE()

# Replace infinite values with NaN
X = X.replace([np.inf, -np.inf], np.nan)
# Fill any remaining NaN values with 0 before applying SMOTE
X_processed = X.fillna(0)

# Transform the dataset
X,y=oversample.fit_resample(X_processed,y)
```

```python
y.value_counts()
```

```python
from sklearn.model_selection import cross_val_score,train_test_split
def classify(model,X,y):
  x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
  # train the model
  model.fit(x_train,y_train)
  return model.score(x_test,y_test)*100
```

Apply Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
LinearReg_acc=classify(model, X, y)
LinearReg_acc
```

Apply Support Vector Machine

```python
from sklearn.svm import SVC
model = SVC(kernel='rbf', random_state = 1)
SVM_acc=classify(model, X, y)
SVM_acc
```

Apply Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
DecTree_acc=classify(model, X, y)
DecTree_acc
```

Apply Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier()
RanFor_acc=classify(model, X, y)
RanFor_acc
```

Apply K.Nearest Neighbours

```python

from sklearn.neighbors  import KNeighborsClassifier
model=KNeighborsClassifier()
knn_acc=classify(model,X,y)
knn_acc
```

```python
Accuracy = [LinearReg_acc, DecTree_acc,RanFor_acc,SVM_acc,knn_acc]
models = ['LogisticRegression', 'DecisionTreeClassifier' , 'RandomForestClassifier', 'Support Vector Machine','KNeighborsClassifier']
```

```python
sns.barplot(x=Accuracy, y=models, color="g")
plt.xlabel('Accuracy in %')
plt.title('Accuracy')
plt.show()
```
